```python
import pandas as pd
df = pd.read_csv("/content/Mall_Customers.csv")
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# 1. Extract the features (fixing the '1-1oo' typo)
X = df[['Annual Income (k$)', 'Spending Score (1-100)']]

# 2. Scale the data to define X_scaled_df
scaler = StandardScaler()
X_scaled_df = scaler.fit_transform(X)

# 3. Calculate WCSS for 1 to 10 clusters
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init="k-means++", random_state=42)
    kmeans.fit(X_scaled_df)
    wcss.append(kmeans.inertia_)

# 4. Plot the Elbow Method graph
plt.figure(figsize=(12, 6))
plt.grid()
plt.plot(range(1, 11), wcss, linewidth=2, color="red", marker="8")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS (Within-Cluster Sum of Squares)")
plt.title("Elbow Method For Optimal K")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# 1. Load data and preprocess
df = pd.read_csv("/content/Mall_Customers.csv")
X = df[["Annual Income (k$)", "Spending Score (1-100)"]]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 2. Fit K-Means
kmeans = KMeans(n_clusters=5, init="k-means++", random_state=42)
kmeans.fit(X_scaled)

# 3. Plot results
plt.figure(figsize=(10, 5))

plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=kmeans.labels_,
    marker="x",
    label="datapoint",
)

plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.title("K-Means Clustering")

plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    marker="*",
    s=300,
    color="red",
    label="centroids",
)

plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data = {"Annual Income": [15,16,17,18,19,60,62,65,67,70],
        "Spending Score": [39,42,45,40,43,70,65,72,68,75]
        }
df = pd.DataFrame(data)
df.head()
```

```python
from sklearn_extra.cluster import KMedoids

# 1. Initialize the model
kmedoids = KMedoids(n_clusters=2, random_state=42)

# 2. Fit the model
kmedoids.fit(df)

# 3. Assign cluster labels
df['Cluster'] = kmedoids.labels_
print(df)

# 4. Display the medoids
print("\nMedoids:")
print(df.iloc[kmedoids.medoid_indices_])
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import MiniBatchKMeans
from sklearn.datasets import make_blobs

# Generate synthetic data
X, _ = make_blobs(n_samples=1000, centers=3, n_features=2, random_state=2)

# Configure and fit MiniBatchKMeans
mbk_means = MiniBatchKMeans(n_clusters=3, batch_size=100, random_state=42)
mbk_means.fit(X)

# Visualize the results
plt.scatter(X[:, 0], X[:, 1], c=mbk_means.labels_, cmap='viridis')
plt.scatter(
    mbk_means.cluster_centers_[:, 0],
    mbk_means.cluster_centers_[:, 1],
    s=200,
    c='red',
    marker='x',
)
plt.title('Mini-batch K-Means Clustering')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import urllib.request
url = "https://raw.githubusercontent.com/tirthajyoti/Machine-Learning-with-Python/master/Datasets/Mall_Customers.csv"
urllib.request.urlretrieve(url, 'Mall_Customer.csv')
data = pd.read_csv('Mall_Customer.csv')
X = data[['Annual Income (k$)', 'Spending Score (1-100)']]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

kMeans = KMeans(n_clusters=5,
                init="k-means++",
                n_init="auto",
                random_state=42)
komeans_fit = kMeans.fit(X_scaled)

plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=komeans_fit.labels_, cmap='viridis')
plt.title('Mall Customer Segmentation (K-Means)')
plt.xlabel('Annual Income (Scaled)')
plt.ylabel('Spending Score (Scaled)')
plt.show()
```
